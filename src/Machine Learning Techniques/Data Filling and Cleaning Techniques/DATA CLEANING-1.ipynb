```python
# drop rows having any NaN or missing value
df3_drop_row=df2_drop_col.dropna()
```

```python
num_var=['MSSubClass', 'LotArea', 'OverallQual', 'OverallCond',
       'YearBuilt', 'YearRemodAdd', 'MasVnrArea', 'BsmtFinSF1', 'BsmtFinSF2',
       'BsmtUnfSF', 'TotalBsmtSF', '1stFlrSF', '2ndFlrSF', 'LowQualFinSF',
       'GrLivArea', 'BsmtFullBath', 'BsmtHalfBath', 'FullBath', 'HalfBath',
       'BedroomAbvGr', 'KitchenAbvGr', 'TotRmsAbvGrd', 'Fireplaces',
       'GarageYrBlt', 'GarageCars', 'GarageArea', 'WoodDeckSF', 'OpenPorchSF',
       'EnclosedPorch', '3SsnPorch', 'ScreenPorch', 'PoolArea', 'MiscVal',
       'MoSold', 'YrSold', 'SalePrice']
# 36= num var (9,4)

plt.figure(figsize=(25,25))
for i,var in enumerate(num_var):
    plt.subplot(9,4,i+1)
    sns.distplot(df[var],color="g", bins=20)
    sns.distplot(df3_drop_row[var],color="b", bins=20)
    
```

```python
# select columns having char dataset only
df3_drop_row.select_dtypes(include=['object']).columns
```

```python
# df['MSZoning'].value_counts () -> counts types present in the MSZoning row
df['MSZoning'].value_counts()
```

```python
# find percentage of dataset counted in above cell
df['MSZoning'].value_counts()/df3_drop_row.shape[0]*100, df3_drop_row['MSZoning'].value_counts()/df3_drop_row.shape[0]*100
```

```python
# concat above data
# row=1 -> concat row wise
pd.concat([df['MSZoning'].value_counts()/df.shape[0]*100, 
           df3_drop_row['MSZoning'].value_counts()/df3_drop_row.shape[0]*100], axis=1,
         keys=['MSZ original', 'MSZ clean'])
```

```python
def cat_var_dist(var):
    return(pd.concat([df[var].value_counts()/df.shape[0]*100, 
           df3_drop_row[var].value_counts()/df3_drop_row.shape[0]*100], axis=1,
         keys=[var+' original', var+' clean']))
    
```

```python
# call above function for MSZoning
cat_var_dist('MSZoning')
```
